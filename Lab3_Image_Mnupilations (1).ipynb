```python
# Import the library
from PIL import Image, ImageOps
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
# ── Load image ────────────────────────────────────

img = data.camera()

plt.imshow(img, cmap="gray")
plt.axis("off")
plt.show()
```

### 1. Scale (increase size)
Double the image dimensions using `Image.resize()` with high-quality resampling `(LANCZOS)`.

```python
# Get the size of the image
img = Image.fromarray(data.camera())
width, height = img.size

# Scaling factors
scale_x = 2
scale_y = 2
```

```python
# ── 1. Increase size (scale ×2) ────────────────────
new_width = int(width * scale_x)
new_height = int(height * scale_y)

# Resize the image using PIL's built-in method

scaled_img = img.resize(
    (new_width, new_height),
    Image.Resampling.LANCZOS
)
```

```python

# Save the scaled image and print the sizes (The new image name should be "task1_1_scaled.jpg")
# Save the scaled image
scaled_img.save("task1_1_scaled.jpg")

# Print the sizes
print("Original size:", img.size)
print("Scaled size:", scaled_img.size)
```

 non-uniform scale (cx=2, cy=1) → stretch horizontally only

```python
cx = 2
cy = 1
```

```python
new_width = int(width * cx)
new_height = int(height * cy)

stretched_img = img.resize(
    (new_width, new_height),
    Image.Resampling.LANCZOS
)
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(stretched_img)
plt.title("Non-uniform Scale")
plt.axis("off")

plt.show()
```

### 2. Rotate 120°
Rotate the image by 120 degrees, expanding the canvas to fit the full rotated image.

### 3. Shear

```python
# -- c. Get the image dimensions ────────────────────
width, height= img.size
```

```python
# -- d. define the shear matrix ──────────────────────────
# Choose X-axis or Y-axis shear. The shear factor controls how much the image slants — start with 0.5 then experiment.
shear_factor = 0.5

shx = shear_factor
shy = 0
canvas_width = int(width * (1 + abs(shear_factor)))
canvas_height = height
```

```python
# -- e. Apply the shear transformation to the image ──────────────────────────
# PIL's transform() takes the inverse affine matrix:
# [ 1    shx   tx ]
# [ shy  1     ty ]
sheared_img = img.transform(
    (canvas_width, canvas_height),
    Image.Transform.AFFINE,
    (1, -shear_factor, 0, 0, 1, 0),
    resample=Image.Resampling.BICUBIC
)
```

```python
# -- f. Save the sheared image(The new image name should be "task1_3_sheared.jpg")
sheared_img.save("task1_3_sheared.jpg")

print("Original size:", img.size)
print("Sheared size:", sheared_img.size)



plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sheared_img)
plt.title("Shear")
plt.axis("off")

plt.show()
```

# Intensity Transformations
Negative · Log · Power Law (Gamma)

```python

# ── 1. Negative ────────────────────────────────────
# Method 1: NumPy array manipulation


arr = np.asarray(img)

negative_arr = 255 - arr

negative_img = Image.fromarray(negative_arr.astype(np.uint8))

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(negative_img)
plt.title("Negative Image")
plt.axis("off")

plt.show()
```

```python
# Method 2: PIL's ImageOps

negative_img_pil = ImageOps.invert(img)

plt.figure(figsize=(5, 5))
plt.imshow(negative_img_pil)
plt.title("Negative using ImageOps")
plt.axis("off")
plt.show()
```

```python
# ── 2. Log transformation ──────────────────────────

# s = c · log(1 + r) --> We need to find c.
# We want the maximum output value (s) to be 255 when the maximum input value (r) is 255:
# 255 = c · log(1 + 255)
# c = 255 / log(1 + 255)



# Apply the log transformation to each pixel
arr = np.asarray(img).astype(np.float32)

# Find c
c = 255 / np.log(1 + 255)

# Apply the log transformation to each pixel
log_transformed_arr = c * np.log(1 + arr)

# Convert values to valid 8-bit range
log_transformed_arr = np.clip(log_transformed_arr, 0, 255)

# Convert back to uint8
log_transformed_arr = log_transformed_arr.astype(np.uint8)

log_img = Image.fromarray(log_transformed_arr)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(log_img)
plt.title("Log Transformation")
plt.axis("off")

plt.show()
```

```python

# ── 3. Power-law / Gamma correction ───────────────
arr = np.asarray(img).astype(np.float32)

# Normalize pixel values to [0, 1]
normalized = arr / 255.0

gamma = 0.5
c = 1

gamma_corrected = c * (normalized ** gamma)

# Convert back to [0, 255]
gamma_corrected = gamma_corrected * 255

gamma_corrected = np.clip(gamma_corrected, 0, 255)
gamma_corrected = gamma_corrected.astype(np.uint8)

gamma_img = Image.fromarray(gamma_corrected)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gamma_img)
plt.title(f"Gamma Correction (γ={gamma})")
plt.axis("off")

plt.show()
```
